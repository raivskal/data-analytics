```python
%%writefile math_server.py
import os
from mcp.server.fastmcp import FastMCP

mcp = FastMCP("Math")

@mcp.tool()
def add(a: int, b: int) -> int:
    return a + b

@mcp.tool()
def multiply(a: int, b: int) -> int:
    return a * b

if __name__ == "__main__":
    # Configure host/port via env vars
    os.environ["MCP_HOST"] = "127.0.0.1"
    os.environ["MCP_PORT"] = "8765"
    mcp.run("sse")
```

```python
%%writefile math_server.py
import os
from mcp.server.fastmcp import FastMCP

mcp = FastMCP("Math")

@mcp.tool()
def add(a: int, b: int) -> int:
    return a + b

@mcp.tool()
def multiply(a: int, b: int) -> int:
    return a * b

if __name__ == "__main__":
    # Use env vars for host/port
    os.environ["MCP_HOST"] = "0.0.0.0"   # listen on all interfaces
    os.environ["MCP_PORT"] = "8765"
    mcp.run("sse")
```
